Use DuckDB to create some dataframes with test data, and write that to a Delta table using the deltalake package, the following code created a basic Delta table containing 10 rows, split across two partitions that we added in two separate steps. 

```python
import duckdb
from deltalake import write_deltalake

con = duckdb.connect()
df1 = con.query(
    "SELECT i AS id, i % 2 AS part, 'value-' || i AS value FROM range(0, 5) tbl(i)"
).df()
df2 = con.query(
    "SELECT i AS id, i % 2 AS part, 'value-' || i AS value FROM range(5, 10) tbl(i)"
).df()
write_deltalake(f"./my_delta_table", df1, partition_by=["part"])
write_deltalake(f"./my_delta_table", df2, partition_by=["part"], mode="append")
```

```python
con.sql("SELECT * FROM delta_scan('./my_delta_table') ORDER BY id;")
```

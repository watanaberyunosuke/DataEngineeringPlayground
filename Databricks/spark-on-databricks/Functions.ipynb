```
netflix_data = spark.read.format("csv") \
    .option("inferSchema", "True") \
    .option("header", "True") \
    .option("sep", ",") \
    .load("dbfs:/FileStore/shared_uploads/harrybwatson@protonmail.com/netflix_list.csv")

display(netflix_data)
```

String manipulation functions

```
from pyspark.sql.functions import initcap, upper, lower

display(netflix_data.select(upper('title'), lower('genres')))
```

```
display(netflix_data.select('title', 'type', initcap('type')))
```

Regular Expressions

```
from pyspark.sql.functions import regexp_replace

display(netflix_data.select('title', 'origin_country', regexp_replace('language', 'En*', 'en')))
```

Replace individual characters

```
from pyspark.sql.functions import translate

display(netflix_data.select('title', translate('title', 'io', '10')))
```

Create temporary views

```
netflix_data.createOrReplaceTempView('netflix_data_table')
```

```
display(spark.sql('select title, rating, genres from netflix_data_table'))
```

```
display(spark.sql("select lower(title), upper(genres) from netflix_data_table"))
```

Numerical Operations

```
display(spark.sql("select title, rating, round(rating) as rounded_rating from netflix_data_table"))
```

Statistical Opertrations

```
netflix_data.stat.corr("rating", "numVotes")
```

```
netflix_data.stat.cov('rating', 'numVotes')
```

```
display(netflix_data.select("rating", "numVotes"))
```
